# Python Assignment 1: Data Structures - Strings & Tuples

## 1. String Concatenation

```python
string1 = "Hello "
name = input("Enter your Name: ")
string2 = name

greeting = string1 + string2
print(greeting)
```

```python
string3 = ", welcome to Python programming"

full_string = greeting + string3
print(full_string)
```

## 2. String Slicing and Indexing

```python
text = full_string

# a. First character
print("a.", text[0])

# b. Last character
print("b.", text[-1])

# c. First 5 characters
print("c.", text[:5])

# d. Last 11 characters
print("d.", text[-11:])

# e. String in reverse
print("e.", text[::-1])

# f. Print the word "Python" using slicing
start = text.find("Python")
print("f.", text[start:start + 6])
```

## 3. String Methods

```python
strM = "Python beginner tutorial"

# a. Uppercase
print("a.", strM.upper())

# b. Lowercase
print("b.", strM.lower())

# c. Capitalize -> back to original form
print("c.", strM.lower().capitalize())

# d. Count of character 't'
print("d.", strM.count("t"))

# e. Replace "Python" with "Machine Learning"
print("e.", strM.replace("Python", "Machine Learning"))
```

## Tuples (Creation, Modification and Access)

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# a. Concatenate the two tuples
t_combine = tuple1 + tuple2
print("a.", t_combine)

# b. Repeat the elements 3 times
print("b.", t_combine * 3)

# c. Access the 3rd element
print("c.", t_combine[2])

# d. First three elements
print("d.", t_combine[:3])

# e. Last three elements
print("e.", t_combine[-3:])
```
